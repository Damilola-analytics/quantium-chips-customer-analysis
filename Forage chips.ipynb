## Data cleaning

```python
transaction["PROD_NAME"] = transaction["PROD_NAME"].str.strip()
```

```python
transaction["PROD_NAME"] = transaction["PROD_NAME"].str.replace(r"\s+"," ",regex = True)
```

```python
transaction = transaction.drop_duplicates()
```

```python
transaction = transaction.dropna()
```

```python
# keep only product containing "chips"

transaction = transaction[transaction["PROD_NAME"].str.contains("chips",case = False,na = False)]
```

```python
## Trim and Clean Data
```

```python
customer["LIFESTAGE"] = customer["LIFESTAGE"].str.strip()
```

```python
customer["LIFESTAGE"] = customer["LIFESTAGE"].str.replace(r"\s+"," ",regex = True)
```

```python
customer["LIFESTAGE"] = customer["LIFESTAGE"].str.title()
```

## Feature Engineering

```python
# Extract pack size
transaction["PACK_SIZE"] = transaction["PROD_NAME"].str.extract("(\d+)").astype(int)
```

```python
# Extract Brand name
transaction["BRAND"] = transaction["PROD_NAME"].str.split().str[0]
```

## Data Visulisation

```python
brand_sales = merge_data.groupby("BRAND")["TOT_SALES"].sum().sort_values(ascending=False)
```

```python
brand_sales.plot(kind="bar",figsize=(10,5))
plt.title("Total Sales by Brand")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

Thins generated the highest total sales, making it the best performing brand. Then followed by Doritos, Cobs, Smiths also contributed significantly to total revenue.
WW and French generated the lowest sales, indicating weaker customer demand. WW and French respectively.
Thins is the strongest-performing brand in terms of revenue and may deserve prominent shelf placement and continued promotional support.

## Strategic Recommendation
Prioritize shelf space, promotions, and inventory for Thins and Doritos, as they are the strongest revenue drivers.
Investigate why WW and French underperform. Consider promotional discounts, product repositioning, and reducing inventory if sales remain consistently low.

```python
segment_sales = merge_data.groupby("LIFESTAGE")["TOT_SALES"].sum()
segment_sales.plot(kind="bar",figsize=(10,5))
plt.title("Total Sales by Lifestage")
plt.xticks(rotation=30)
plt.show()
```

### Interpretation
Older Singles/Couples generated the highest tottal sales among all customer lifestages. This indicates that this segment contributes signifucantly to chips sales and could be targeted wit family-size promotions and value packs.
Retirees and Older Families also contributed substantially. New Families contributed the least to overall revenue.

## Strategic Recommendation
Focus marketing campaigns and loyalty programs on Older Singles/Couples, Retirees, and Older Families, since they generate the largest share of revenue.
Design targeted promotions for New Families to encourage trial purchases and increase engagement


```python
Premium_Customer_Sales = merge_data.groupby("PREMIUM_CUSTOMER")["TOT_SALES"].sum()
Premium_Customer_Sales.plot(kind="bar",figsize=(8,4))
plt.title("Total Sales by Premium Customers")
plt.xticks(rotation=30)
plt.show()
```

### Interpretation

1)Mainstream customers generated the highest total sales, indicating they purchase chips more frequently and in greater quantities than other customer groups.

2)Budget customers contributed the second-highest total sales, showing they are also an important customer segment despite being more price-conscious.

3)Premium customers generated the lowest total sales, suggesting they purchase less often, buy fewer packs per transaction, and represent a smaller customer base.

### Strategic Recommendation

The company should continue prioritizing Mainstream customers because they generate the highest revenue. At the same time, further analysis should be conducted on Premium customers to identify opportunities for increasing their purchase frequency or encouraging higher basket sizes through premium product bundles and targeted promotions.


```python
#Frequent Customer who purchase chips most
Most_Purchase_Customer = merge_data.groupby("LIFESTAGE")["TXN_ID"].nunique()

Most_Purchase_Customer.plot(kind="bar",figsize=(8,4))
plt.title("Volume of Transaction per chips purchase")
plt.xticks(rotation=30)
plt.show()
```

## Interpretation
The analysis indicates that Older Singles/Couples have the highest number of chip purchase transactions, followed by Retirees, Older Families, and Young Families. These customer segments purchase chips more frequently and contribute the largest share of transaction volume.
In contrast, New Families recorded the lowest number of transactions, suggesting they purchase chips less frequently than other customer segments. This reflect different shopping priorities, lower demand for snack products.

## Strategic Recommendations
Retain high-frequency customers. Continue targeting Older Singles/Couples, Retirees, Older Families, and Young Families with loyalty rewards, personalized promotions, and consistent availability of their preferred chip brands and pack sizes.
Increase engagement with New Families. Introduce family-oriented promotions, discounts, or bundled snack offers to encourage more frequent purchases. Cross-promoting chips alongside grocery staples may also help increase transaction frequency.
Tailor marketing by segment. Avoid a one-size-fits-all approach. Focus retention efforts on high-performing segments while using targeted campaigns to stimulate demand among New Families.

```python
# Sales trend overtime
merge_data["Month"] = merge_data["DATE"].dt.to_period("M")
monthly_sales = merge_data.groupby("Month")["TOT_SALES"].sum()

monthly_sales.plot(figsize=(10,5), marker="o")
plt.title("Monthly Sales Trend")
plt.grid(True)
plt.show()
```

## Interpretation
Sales fluctuate throughout the year.
Sales peak in December, likely due to holiday shopping.
Sales decline sharply in February before recovering in March.

## Strategic Recommendation
Increase stock levels and promotional activities before December to capitalize on seasonal demand.
Introduce special promotions during lower-performing months such as February to stabilize sales.


```python
Pack_Size_Performance = merge_data.groupby("PACK_SIZE")["TOT_SALES"].sum().sort_values(ascending=False)
Pack_Size_Performance.plot(kind="bar",figsize=(10,5))
plt.title("Pack Size Per Revenue")
plt.xticks(rotation=30)
plt.show()
```

175g and 110g packs also performed well.
Larger pack sizes such as 330g generated considerably less revenue.

Prioritize production and shelf placement for the most popular pack sizes (170g and 175g).

```python
# Brand VS LifeStage
brand_lifestage = merge_data.pivot_table(values="TOT_SALES", index="LIFESTAGE", columns ="BRAND", aggfunc="sum", fill_value=0)

sns.heatmap(brand_lifestage, annot=True, fmt=".0f", cmap="YlGnBu", linewidths=0.5)

plt.title("Brand VS customer Life Stage")
plt.xlabel("Brand")
plt.ylabel("Customer Life Stage")

plt.tight_layout()
plt.show()
```

## Interpretation
Thins consistently recorded high sales across almost every customer life stage.
Older Singles/Couples, Retirees, and Older Families purchased the highest quantities across multiple brands.
New Families showed relatively low purchasing activity for all brands.

## Strategic Recommendation
Tailor marketing campaigns by customer segment.
Promote Thins heavily to high-value customer groups, especially Older Singles/Couples and Retirees.
Develop targeted campaigns aimed at increasing brand awareness among New Families.
